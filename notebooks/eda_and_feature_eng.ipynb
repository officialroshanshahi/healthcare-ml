```python
age_bins = [0, 18, 40, 65, 120]

age_labels = ['Pediatric', 'Young_Adult', 'Middle_Aged', 'Senior']

df_2['age_group'] = pd.cut(df_2['age'], bins=age_bins, labels=age_labels)

df_2 = df_2.drop(columns=['age'])
```

```python
# using category_encoders for hospital to find new features
from category_encoders import TargetEncoder
from sklearn.model_selection import train_test_split


df_2['date_of_admission'] = pd.to_datetime(df_2['date_of_admission'])
df_2['discharge_date'] = pd.to_datetime(df_2['discharge_date'])


y_res = df_2['test_results'].map({'Normal': 0, 'Inconclusive': 1, 'Abnormal': 2}).values
y_dur = (df_2['discharge_date'] - df_2['date_of_admission']).dt.days.values.astype(np.float32)
y_bil = df_2['billing_amount'].values.astype(np.float32)

# dropping unnecessary columns
X = df_2.drop(columns=['name', 'doctor','discharge_date', 'test_results', 'billing_amount', "room_number"])

X['admission_month'] = X['date_of_admission'].dt.month
X['admission_dayofweek'] = X['date_of_admission'].dt.dayofweek

X.drop(columns=['date_of_admission'], inplace=True)


X_train, X_test, y_tr_res, y_te_res = train_test_split(
    X, y_res, test_size=0.20, random_state=42
)


_, _, y_tr_dur, y_te_dur = train_test_split(
    X, y_dur, test_size=0.20, random_state=42
)


_, _, y_tr_bil, y_te_bil = train_test_split(
    X, y_bil, test_size=0.20, random_state=42
)

# smoothing 10 , 50, 100 was causing high correlation so using smoothing 1000 but even 1000 wasn't able to lower
# the correlation from 0.9 so i decided to drop hospital itself
te_tr = TargetEncoder(cols=['hospital'], smoothing=1000.0)
te_dur = TargetEncoder(cols=['hospital'], smoothing=1000.0)
te_bil = TargetEncoder(cols=['hospital'], smoothing=1000.0)


X_train['hospital_result_'] = te_tr.fit_transform(X_train['hospital'], y_tr_res)
X_test['hospital_result_'] = te_tr.transform(X_test['hospital'])


X_train['hospital_dur_'] = te_dur.fit_transform(X_train['hospital'], y_tr_dur)
X_test['hospital_dur_'] = te_dur.transform(X_test['hospital'])


X_train['hospital_bill_'] = te_bil.fit_transform(X_train['hospital'], y_tr_bil)
X_test['hospital_bill_'] = te_bil.transform(X_test['hospital'])

print(df.groupby('hospital')[['billing_amount', 'discharge_date']].mean())

X_train.drop(columns=['hospital'], inplace=True)
X_test.drop(columns=['hospital'], inplace=True)

X_train
```

```python
categorical_cols = ["gender", "blood_type", "medical_condition", "insurance_provider", "admission_type", "medication", "age_group"]

# for lightgbm
X_train_wo_enc = X_train.copy()
X_test_wo_enc = X_test.copy()

for col in categorical_cols:
    X_train_wo_enc[col] = X_train_wo_enc[col].astype('category')
    X_test_wo_enc[col] = X_test_wo_enc[col].astype('category')

X_train = pd.get_dummies(X_train, columns=categorical_cols, drop_first=True, dtype=int)
X_test = pd.get_dummies(X_test, columns=categorical_cols, drop_first=True, dtype=int)

X_train.head()
```

```python
# DONT RUN THIS TO GO DIRECTLY TO CLASSIFIER ONLY MODE

import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from sklearn.metrics import (accuracy_score, f1_score, mean_absolute_error, precision_score, 
                             recall_score, roc_auc_score,
                             root_mean_squared_error, r2_score)
from sklearn.model_selection import RandomizedSearchCV

sns.set_theme(style="whitegrid")

save_dir = "../plots/evaluate/v1/models/"
os.makedirs(save_dir, exist_ok=True)

best_models = {}

# Convert target variables directly to 1D arrays on their original scale
y_tr_res = y_tr_res.values.ravel() if hasattr(y_tr_res, 'values') else np.asarray(y_tr_res).ravel()
y_te_res = y_te_res.values.ravel() if hasattr(y_te_res, 'values') else np.asarray(y_te_res).ravel()

y_tr_bil = y_tr_bil.values.ravel() if hasattr(y_tr_bil, 'values') else np.asarray(y_tr_bil).ravel()
y_te_bil = y_te_bil.values.ravel() if hasattr(y_te_bil, 'values') else np.asarray(y_te_bil).ravel()

y_tr_dur = y_tr_dur.values.ravel() if hasattr(y_tr_dur, 'values') else np.asarray(y_tr_dur).ravel()
y_te_dur = y_te_dur.values.ravel() if hasattr(y_te_dur, 'values') else np.asarray(y_te_dur).ravel()

for model_name, sub_dict in model_dict.items():
    best_models[model_name] = {}
    
    for target_key, sub_model in sub_dict.items():
        print(f"--- Tuning & Training {model_name} for target: {target_key} ---")
        
        if model_name == 'LGBMClassifier':
            X_train_curr = X_train_wo_enc
            X_test_curr  = X_test_wo_enc
            feature_names = X_train_wo_enc.columns.tolist() if hasattr(X_train_wo_enc, 'columns') else [f'Feat_{i}' for i in range(X_train_wo_enc.shape[-1])]
        else:
            X_train_curr = X_train
            X_test_curr  = X_test
            feature_names = X_train.columns.tolist() if hasattr(X_train, 'columns') else [f'Feat_{i}' for i in range(X_train.shape[-1])]
        
        param_dist = model_params[model_name]
        
        if 'Classifier' in model_name:
            search = RandomizedSearchCV(
                estimator=sub_model,
                param_distributions=param_dist,
                n_iter=50,
                scoring='accuracy',
                cv=5,
                random_state=42,
                verbose=1
            )
            
            search.fit(X_train_curr, y_tr_res)
            model = search.best_estimator_
            
            y_pred = model.predict(X_test_curr)
            y_probs = model.predict_proba(X_test_curr) if hasattr(model, "predict_proba") else model.decision_function(X_test_curr)
            
            f1 = f1_score(y_te_res, y_pred, average='macro')
            precision = precision_score(y_te_res, y_pred, average='macro')
            recall = recall_score(y_te_res, y_pred, average='macro')
            roc_auc = roc_auc_score(y_te_res, y_probs, multi_class='ovr', average='macro')

            best_models[model_name][target_key] = {
                'model': model,
                'cv_score': search.best_score_,
                'test_metrics': {
                    'accuracy': accuracy_score(y_te_res, y_pred),
                    'f1': f1,
                    'roc_auc': roc_auc,
                    'precision': precision,
                    'recall': recall
                }
            }

        else:
            y_train_curr = y_tr_dur if 'dur' in target_key else y_tr_bil
            y_test_orig  = y_te_dur if 'dur' in target_key else y_te_bil
            
            search = RandomizedSearchCV(
                estimator=sub_model,
                param_distributions=param_dist,
                n_iter=50,
                scoring='neg_root_mean_squared_error',
                cv=5,
                random_state=42,
                verbose=1
            )
            search.fit(X_train_curr, y_train_curr)
            model = search.best_estimator_
            
            y_pred = model.predict(X_test_curr)
            
            best_models[model_name][target_key] = {
                'model': model,
                'cv_score': search.best_score_,
                'test_metrics': {
                    'rmse': root_mean_squared_error(y_test_orig, y_pred),
                    'mae': mean_absolute_error(y_test_orig, y_pred),
                    'r2': r2_score(y_test_orig, y_pred)
                }
            }

        importances = None
        if hasattr(model, 'feature_importances_'):
            importances = model.feature_importances_
        elif hasattr(model, 'coef_'):
            importances = np.abs(model.coef_ if model.coef_.ndim > 1 else model.coef_)
            
        if importances is not None:
            feat_imp_df = pd.DataFrame({
                'Feature': feature_names,
                'Importance': importances
            }).sort_values(by='Importance', ascending=False).head(15)
            
            plt.figure(figsize=(10, 5))
            sns.barplot(
                data=feat_imp_df, 
                x='Importance', 
                y='Feature', 
                palette='viridis',
                hue='Feature',
                legend=False
            )
            plt.title(f"Top 15 Features: {model_name} ({target_key})", fontsize=12, fontweight='bold')
            plt.xlabel("Importance Score")
            plt.ylabel("Features")
            plt.tight_layout()
            
            filename = f"{model_name.lower()}_{target_key.lower()}.png"
            file_path = os.path.join(save_dir, filename)
            
            plt.savefig(file_path, dpi=300, bbox_inches='tight')
            print(f" Saved plot to: {file_path}")
            
            plt.show()
            plt.close()
        else:
            print(f"Feature importances not natively supported for {model_name}.\n")
```

```python
import torch.nn as nn
import torch
from skorch import NeuralNetClassifier
from sklearn.preprocessing import StandardScaler


scaler = StandardScaler()

continuous_features = ['billing_amount', 'duration', 'admission_month', 'admission_dayofweek']
binary_features = [col for col in X_train.columns if col not in continuous_features and col != 'test_results']


X_train_con = pd.DataFrame(
    scaler.fit_transform(X_train[continuous_features]), 
    columns=continuous_features,
    index=X_train.index
)
X_test_con = pd.DataFrame(
    scaler.transform(X_test[continuous_features]), 
    columns=continuous_features,
    index=X_test.index
)

X_train_scaled = pd.concat([X_train_con, X_train[binary_features]], axis=1)
X_test_scaled  = pd.concat([X_test_con, X_test[binary_features]], axis=1)

y_train_encoded = (y_tr_res.values if hasattr(y_tr_res, 'values') else y_tr_res).astype('int64').squeeze()
y_test_encoded = (y_te_res.values if hasattr(y_te_res, 'values') else y_te_res).astype('int64').squeeze()


class ResultClassifier(nn.Module):
    def __init__(self, input_dim):
        super().__init__()
        self.shared = nn.Sequential(
            nn.Linear(input_dim, 256),
            nn.BatchNorm1d(256),
            nn.ReLU(),
            nn.Dropout(0.3),
            nn.Linear(256, 128),
            nn.BatchNorm1d(128),
            nn.ReLU(),
            nn.Dropout(0.2)
        )
        self.out_res = nn.Linear(128, 3)
        
    def forward(self, x):
        h = self.shared(x)
        return self.out_res(h)


nn_wrapper = NeuralNetClassifier(
    ResultClassifier,
    module__input_dim=X_train_scaled.shape[1],
    criterion=nn.CrossEntropyLoss,
    optimizer=torch.optim.Adam,
    max_epochs=20,
    iterator_train__batch_size=128,
    iterator_train__shuffle=True,
    verbose=0
)


nn_params = {
    'lr': [0.05, 0.1, 0.2, 0.3],
    'iterator_train__batch_size':[32, 64, 128, 256], #
    'max_epochs': [20, 30, 50, 60, 80]    # checking if model needs more processing time  
}


search = RandomizedSearchCV(
    estimator=nn_wrapper,
    param_distributions=nn_params,
    cv=5,
    n_iter=10,
    verbose=1,
    error_score='raise'
)

print("Starting hyperparameter optimization...")

search.fit(X_train_scaled.values.astype('float32'), y_train_encoded)

print(f"Optimal parameters found: {search.best_params_}")

best_model = search.best_estimator_
history = best_model.history

train_loss = history[:, 'train_loss']
valid_loss = history[:, 'valid_loss']
valid_acc = history[:, 'valid_acc']
epochs = range(1, len(train_loss) + 1)

plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.plot(epochs, train_loss, 'o-', label='Train Loss', color='royalblue')
plt.plot(epochs, valid_loss, 's-', label='Validation Loss', color='darkorange')
plt.title('Model Loss Performance')
plt.xlabel('Epochs')
plt.ylabel('Loss Value')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.subplot(1, 2, 2)
plt.plot(epochs, valid_acc, 'g^-', label='Validation Accuracy')
plt.title('Validation Accuracy Evolution')
plt.xlabel('Epochs')
plt.ylabel('Accuracy Score')
plt.grid(True, linestyle='--', alpha=0.6)
plt.legend()

plt.tight_layout()

plot_save_path = os.path.join(output_dir, "new_clf_nn_metrics_plot.png")
plt.savefig(plot_save_path, dpi=300)

plt.show()

plt.close()

print(f"Metrics performance plot successfully generated and saved to: {plot_save_path}")
```
